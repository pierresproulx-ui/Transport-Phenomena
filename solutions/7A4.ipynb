#### Exercice 7A4, Transport Phenomena


#### Solution utilisant le package python fluids


Comme dans le problème 7A2, on a une conduite seulement

><center>
$
  \boxed{ \sum_{entrée} \bigg ( \frac {1}{2} v_1^2 + g h_1 + \frac {p_1}{\rho_1} \bigg)w-\sum_{sortie} \bigg ( \frac {1}{2} v_2^2 + g h_2 + \frac {p_2}{\rho_2} \bigg )w=-W_m +E_c+ w \frac {V^2}{2}  \big [ \sum K_{f-sur-L} +  \sum K_{perturbations}   \big ]} (1)
  $
</center>

En appliquant l'équation à 7A4:

on obtient rapidement
<center>
$\boxed{ \frac  {w}{\rho}  ( p_1 + \rho g h_1-p_2 - \rho g h_2) =- {W_m}  
+ w \frac {V^2}{2}  \big [ \sum K_{f-sur-L} +  \sum K_{perturbations}  \big ]} (2)$
</center>

On aura à utiliser la définition du rayon hydraulique.

Voyons maintenant comment on solutionne avec le package fluids de Caleb Bell.

```python
import thermo as th
from math import *
from fluids.units import *
from thermo.units import Stream 
#
Ri=3*u.inch
Ro=7*u.inch
debit=241*u.US_liquid_gallon/u.min
densite=62.4*u.lb/u.ft**3
surface=pi*(Ro**2-Ri**2)
perimetre=2*pi*Ri+2*pi*Ro
V=debit/surface
print(V.to(u.m/u.sec))
H2=5*u.ft         # C'est la différence de hauteur entre H2 et H1
P1=1*u.atm        # on pourrait mettre P1 et P2 à zéro, l'important c'est qu'elles soit égales
P2=1*u.atm+densite*u.gravity*H2  # Pression hydrostatique:  (rho g (h2 - h1)) dans l'eq. (2)  
Rh=surface/perimetre     # Rayon hydraulique et diamètre hydraulique, Dh=4Rh, attention!!!
Re = Reynolds(D=4*Rh, rho=densite, mu=1.0*u.cP, V=V)
print('Reynolds ',Re)
L=20.3*u.ft
w=densite*V*surface     # ces unités!!! FLUIDS en tient compte, merci beaucoup.
print(w.to(u.lb/u.sec))
K=K_from_f(fd=friction_factor(Re=Re),L=L,D=4*Rh)   # Calculer le terme associé à la conduite droite
K += entrance_sharp()                             # la perte associée à l'entrée
K += exit_normal()                                # la perte associée à la sortie
dp=dP_from_K(K=K,rho=densite,V=V).to(u.Pa)        # traduit K en termes de pression
dp+=(P2-P1)                                 # ajoute aux pressions connues, on obtient dp total
Puissance=(dp*w/densite).to(u.hp)           # il ne reste que la puissance comme inconnue.
print(Puissance,Puissance.to(u.W))          # La puissance dans les unités que l'on veut.
```
