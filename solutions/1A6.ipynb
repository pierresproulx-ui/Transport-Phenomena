```python
# Exercice 1A7
#
import thermo as th
import fluids
from fluids.units import *
```

```python
T=273.16  # juste au-dessus du point de fusion!
eau=th.Chemical('water',T=T)
print("Viscosité de l'eau à T= {:<6.2f} 'K '{:<12.3e}".format(T,(eau.mu*u.Pa*u.s).to(u.cpoise)))
T=373.15
eau.calculate(T=T)
print("Viscosité de l'eau à T= {:<6.2f} 'K '{:<12.3e}".format(T,(eau.mu*u.Pa*u.s).to(u.cpoise)))
```
