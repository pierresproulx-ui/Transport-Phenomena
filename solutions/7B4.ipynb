L'équation de bilan d'énergie mécanique pour les deux conduites est faite en prenant les points 1 et 2 à l'entrée et la sortie des conduites la hauteur est la même, pas de pompe et pas de terme de compression, la vitesse est la même aux points 1 et 2 dans les deux cas:

<center>
$\begin{equation}
  \boxed{  \bigg (  \frac {p_1} {\rho} \bigg )_1 w-\bigg ( \frac {p_2} {\rho} \bigg )_1 w=(\frac  {L}{2R_1} v^2 f_1 ) w}
  \end{equation} $
</center>

### et

<center>
$\begin{equation}
  \boxed{  \bigg (  \frac {p_1} {\rho} \bigg )_2 \frac {w}{2}-\bigg ( \frac {p_2} {\rho} \bigg )_2 \frac {w}{2}=(\frac  {L}{2R_2} v^2 f_2 ) \frac {w}{2}}
  \end{equation} $
</center>

##### Les tanks sont au même niveau, pas de pompes  et les deux débits doivent être les mêmes avec deux tubes ayant chacun la moitié de la surface. 

#### L'équation de bilan ci-haut devient en isolant la différence de pression pour les deux expressions.
### $ \left ( {p_1}- {p_2} \right )=\frac {1}{2} \rho v^2 \frac {L}{R} 0.0791 Re^{-0.25}$

#### qu'on peut poser pour les deux systèmes.

```python
import sympy as sp
from IPython.display import *
sp.init_printing(use_latex=True)
v,rho,L,R1,R2,p1,p2,mu=sp.symbols('v,rho,L,R1,R2,p1,p2,mu')
dp1=(1/2*rho*v*L/R1)**4/(rho*v*(2*R1)/mu)*0.0791
dp2=(1/2*rho*v*L/R2)**4/(rho*v*(2*R2)/mu)*0.0791  # à la puissance 4 pour avoir une solution qui ne fait pas de racine 1/4 pour f
rapport=(dp1/dp2)**0.25    # remet a la puissance originale en prenant ** 1/4
display(rapport.subs(R2,R1/(2**0.5))) 
display(1/2**(5/8))   # pour comparaison avec la solution
```
