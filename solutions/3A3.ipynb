Puisqu'on a la loi des gaz parfaits qui est correcte à une pression faible comme la pression atmosphérique, on peut écrire la densité à partir de celle-ci comme:

$PV=NRT$ devient en isolant la concentration molaire $\frac {N}{V} = \frac {P}{RT}$ et en multipliant par la masse molaire:$M \frac {N}{V}=\rho = M\frac {P}{RT}$ Ensuite, on écrit l'équation générale de qté de mouvement B6-3 pour ce problème particulier, puisque'il n'y a aucune vitesse, les seuls termes restant sont: $0=\frac{dP}{dz}+\rho g_z$ 

```python
R,z,g,H,M,T_0,c=sp.symbols('R,z,g,H,M,T_0,c')
T=sp.Function('T')(z)
p=sp.Function('p')(z)
T=T_0+c*z
rho=p*M/(R*T)
p=sp.dsolve(sp.Eq(sp.diff(p,z),-rho*g))
cinit=sp.Eq(p.rhs.subs(z,0),sp.symbols('p_0'))
constante=sp.solve(cinit,sp.symbols('C1'))
p=p.subs(sp.symbols('C1'),constante[0])
display(p)
air=th.Mixture('air')
Mw=air.MW
DH=2023-602
dico={'T_0':530,'g':32.19,'H':DH,'R':4.9686e4,'c':-0.003}
display(p.subs(z,H).subs(M,Mw).subs(dico).subs(sp.symbols('p_0'),750))
```
