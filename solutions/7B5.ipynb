#### Solution

#### a) Forme simplifiée des équations de bilan d'énergie mécanique
L'équation de bilan d'énergie mécanique pour les deux conduites est faite en prenant les points 1 et 2 à l'entrée et la sortie des conduites:

<center>
$\begin{equation}
  \boxed{ \sum_{entrée} \bigg ( \frac {1}{2} v_1^2 + g h_1 + \frac {p_1}{\rho_1} \bigg)w_1-\sum_{sortie} \bigg ( \frac {1}{2} v_2^2 + g h_2 + \frac {p_2}{\rho_2} \bigg )w_2=-W_m +E_c+ \sum_{i-cond} (w_i 2 \frac  {L}{D} v^2 f)_i  + \sum_{j-pert} (w_i \frac  {1}{2} v^2 e_v)_j}
  \end{equation} $(1)
</center>

Pour des conduites non-circulaires le terme 

<center>
$\boxed{2\frac {L}{D}}$
</center>

est remplacé par 

<center>

$ \boxed{\frac {L}{2R_h}}$
</center>

et le nombre de Reynolds est défini comme 

<center>
$\boxed{\frac {4\rho V R_h} {\mu} }$(1)
</center>

Ainsi pour chacune des deux configurations on obtient en appliquant (1)

<center>
$\begin{equation*}
  \boxed{ \frac {p_1} {\rho} - \frac {p_2} {\rho}=( \frac  {L}  {2R_{h_I}}    v_I   ^2 f_I)}
  \end{equation*}$ (2)
</center>

<center>
$\begin{equation*}
  \boxed{ \frac {p_1} {\rho} - \frac {p_2} {\rho} =( \frac  {L}  {2R_{h_{II}}}    v_{II}   ^2 f_{II})}
  \end{equation*}$ (3)
</center>

#### b)Les deux différences de pression doivent être égales :
<center>
$\begin{equation*}
  \boxed{ ( \frac  {L}  {2R_{h_I}}    v_I   ^2 f_I)=( \frac  {L}  {2R_{h_{II}}}    v_{II}   ^2 f_{II})
         }
  \end{equation*} $(4)
</center>
et en insérant Blasius pour f:
<center>
$\begin{equation*}
  \boxed{ ( \frac  {L}  {2R_{h_I}}    v_I   ^2 (\frac {0.0791 \mu}{4 \rho v_I R_{h_I}})^{1/4}=( \frac  {L}  {2R_{h_{II}}}    v_{II}   ^2 (\frac {0.0791 \mu}{4 \rho v_{II} R_{h_{II}}})^{1/4}}
\end{equation*} $(5)

</center>
simplifions:
<center>
$\begin{equation*}
  \boxed{ ( \frac  {1}  {R_{h_I}}    v_I^2 (\frac {1}{ v_I R_{h_I}})^{1/4}=( \frac  {1}  {R_{h_{II}}}    v_{II}^2 (\frac {1}{v_{II} R_{h_{II}}})^{1/4}}
  \end{equation*} $(6)
</center>
isolons le ratio des vitesses:

<center>
$\begin{equation*}
  \boxed{  \frac  {v_I^{7/4}} {v_{II}^{7/4}} =  \frac { R_{h_I}^{5/4}} {R_{h_{II}}^{5/4}}}
  \end{equation*}$ (7)
</center>

Les deux débits doivent être égaux, donc on a que $v_I A_I=v_{II} A_{II}$:
<center>
$\begin{equation*}
  \boxed{     v_I  H_I W_I = v_{II}  H_{II} W_{II}
         }
  \end{equation*}$
</center>
ou
<center>
$\begin{equation*}
  \boxed{     \frac {v_I}{v_{II}}   = \frac   {H_{II} W_{II}} {H_I W_I}
         }
  \end{equation*}$ (8)
</center>
Substituons dans l'équation 7:
<center>
$\begin{equation*}
  \boxed{ \bigg ( \frac   {H_{II} W_{II}} {H_I W_I} \bigg )^{7/4} =  \bigg (\frac { R_{h_I}} {R_{h_{II}}} \bigg )^{5/4}
         }
  \end{equation*}$ (9)
</center>

La définition des rayons hydrauliques est section/périmètre, pour les deux conduites on a donc:
<center>
$\begin{equation*}
  \boxed{     R_{h_I} = \frac {H_I*W_I}{2*(H_I+W_I)}
         }
  \end{equation*}$ (10)
</center>
et 
<center>
$\begin{equation*}
  \boxed{     R_{h_{II}} = \frac {H_{II}*W_{II}}{2*(H_{II}+W_{II})}
         }
  \end{equation*}$ (11)
</center>
En substituant ces définitions dans (9), finalement:
<center>
$\begin{equation*}
  \boxed{ \bigg ( \frac   {H_{II} W_{II}} {H_I W_I} \bigg )^{7/4} =  \bigg (\frac { \frac {H_I*W_I}{2*(H_I+W_I)} } {\frac {H_{II}*W_{II}}{2*(H_{II}+W_{II})} } \bigg )^{5/4}
         }
  \end{equation*}$ (12)
</center>
Cette dernière ne contient qu'une inconnue, $W_{II}$
#### c) La solution en utilisant le solver de python/scipy

```python
#
# Fonction : c'est l'équation à résoudre, l'équation (11)
#
def equation(WII):
    rh1=(HI*WI)/(2*(HI+WI))
    rh2=(HII*WII)/(2*(HII+WII))
    return ((HII*WII)/(HI*WI))**(1.75)-(rh1/rh2)**(1.25)
#
# Importer l'outil de recherche des zéros d'équations non-linéaires
from scipy.optimize import fsolve
#
# Données numériques
#
HI=4
WI=4
HII=2
#
# Solution
#
WII=fsolve(equation,(1))# fonction dont on cherche le zéro et estimé.
print('WII = ',WII)
```
