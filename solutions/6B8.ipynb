```python
#
# Pierre Proulx
#
# Préparation de l'affichage et des outils de calcul symbolique
#
import sympy as sp
from IPython.display import *
import thermo as th
sp.init_printing(use_latex=True)
# Paramètres, variables et fonctions
rho_s,rho,D,v_inf,mu,g,Re,drho=sp.symbols('rho_s,rho,D,v_inf,mu,g,Re,Delta_rho')  
f=4/3*g*D/v_inf**2*(rho_s-rho)/rho  # equation définissant le facteur f
```

### On peut faire apparaître le nombre de Reynolds dans cette expression en multipliant par le facteur $\frac {\mu^2}{\rho D^3} \frac {\rho D^3}{\mu^2} $ ce qui donnera : $\frac {4}{3} \frac { g \Delta \rho} {Re^2} \frac {\rho D^3}{\mu^2}$ 

```python
f=4/3*g/Re**2*drho*rho*D**3/mu**2  # le facteur f en fonction de Reynolds
display("L'expression f formulée en fonction de Re",f)
solution=sp.solve(f-1,D)
display("Solution de l'expression (attention, il y a 3 racines car c'est un degré 3)",solution[0])
```

```python
dico={'Re':100,'mu':0.01,'Delta_rho':7,'g':981,'rho':1}  # pas en SI, en cgs
Dsolution=solution[0].subs(dico)
display(Dsolution.evalf(4))      # en cm
```
