```python
#
# Pierre Proulx
#
# Préparation de l'affichage et des outils de calcul symbolique
#
import sympy as sp
from IPython.display import *
sp.init_printing(use_latex=True)
```

```python
# Paramètres, variables et fonctions
rho_s,rho,D,v_inf,mu,g=sp.symbols('rho_s,rho,D,v_inf,mu,g')                                                            
```

```python
f=4/3*g*D/v_inf**2*(rho_s-rho)/rho  # equation définissant le facteur f
#display(f)
Re=rho*v_inf*D/mu                           
f_v=(24/Re)        # équation calculant f par la friction, régime de Stokes
#display(f_v)
Force=sp.pi*D**3/6*g*(rho_s-rho)
#display(Force)
```

```python
masse=0.05e-3
volume=5e-3**3*sp.pi/6
densite=masse/volume
v=0.005
#display(densite)
dico={'rho_s':densite,'rho':900,'D':0.005,'g':-9.81,'v_inf':v}
#display(Re.subs(dico))
#display('Force en dynes',Force.subs(dico).evalf()*1e5)
#display('coefficient de friction',f.subs(dico).evalf())
viscosite=sp.solve((24*mu/(D*v_inf*rho))-f,mu)   # valide si Reynolds << 1
#display(viscosite)
display('viscosité en kg/m-sec',viscosite[0].subs(dico).evalf())
display('viscosité en g/cm-sec',viscosite[0].subs(dico).evalf()*10)
vis=viscosite[0].subs(dico).evalf()
dico2={'mu':vis}
display('Reynolds, le calcul de la viscosité est ok si Re<<1',(Re.subs(dico).subs(dico2)))
```
