```python
#
# Exercice 1C1 a)
#
import sympy as sp
from IPython.display import *
sp.init_printing(use_latex=True)
m,k,T,n,u = sp.symbols('m,k,T,n,u',positive=True)
f = sp.Function('f')(u)
f=4*sp.pi*n*u**2*(m/(2*sp.pi*k*T)**(3/2))*sp.exp(-m*u**2/(2*k*T))
N=sp.integrate(u*f,(u,0,sp.oo))
D=sp.integrate(f,(u,0,sp.oo))
ubar=N/D
display(ubar.nsimplify()**2)
```

```python
# Exercice 1C1 c)
#
N1=sp.integrate(1/2*m*u**2*f,(u,0,sp.oo))
D1=sp.integrate(f,(u,0,sp.oo))
equation=sp.Eq(1/2*m*u**2,N1/D1)
display(equation)
```
