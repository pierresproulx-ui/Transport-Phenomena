### Probleme 18B9

```python
#
# Pierre Proulx
#
# Preparation de l'affichage et des outils de calcul symbolique
#
import sympy as sp
from IPython.display import *
sp.init_printing(use_latex=True)
%matplotlib inline
# Parametres, variables et fonctions
z,C_A0,C_Ad,D_AB,S,delta,delta_z,L=sp.symbols('z C_A0 C_Ad D_AB S delta delta_z L')  
N_A=sp.Function('N_A')
C_A=sp.Function('C_A')(z)    
```

```python
#
# Le bilan est identique a la section 18-4 sans la reaction
#
eq=(S*N_A(z)-S*N_A(z+delta_z))/delta_z
display(eq)
```

```python
eq=sp.limit(eq,delta_z,0).doit()
display(eq)
```

```python
fick=-D_AB*sp.diff(C_A,z)
eq=eq.subs(N_A(z),fick)
display(eq)
```

```python
C_A=sp.dsolve(eq,C_A)
display(C_A)
C=C_A                    # je garde une copie de l'expression pour plus tard
```

```python
C_A=C_A.rhs
#
# Pose et solution des 2 equations de conditions aux limites pour C1 et C2
cl1=sp.Eq(C_A.subs(z,0),C_A0)
cl2=sp.Eq(C_A.subs(z,delta),C_Ad)
constantes=sp.solve([cl1,cl2],sp.symbols('C1,C2') )     # pour trouver C1 et C2
display(constantes)
```

```python
C_A_solution=C_A.subs(constantes)
display('Partie a et b, resultat',C_A_solution)
#
# et pour la partie c), la loi de Fick Na=-D_AB  dCa/dz
#
Flux=-D_AB*C_A_solution.diff(z)
display(' Le flux et donne en appliquant la loi de Fick')
display('on a deja obtenu le profil de C_A',Flux.simplify())
```

```python
import matplotlib.pyplot as plt
plt.rcParams['figure.figsize']=10,8
dico={'D_AB':1.e-10,'C_A0':1,'delta':1,'C_Ad':0}
Cp=C_A_solution.subs(dico)
display(Cp)
```
