### Problème 18B1

```python
#
# Pierre Proulx
#
#
# Préparation de l'affichage et des outils de calcul symbolique
#
import sympy as sp
from IPython.display import *
sp.init_printing(use_latex=True)
%matplotlib inline
```

```python
# Paramètres, variables et fonctions

z,z_1,z_2,C,D_AB,X_1,X_2=sp.symbols('z,z_1,z_2,C,D_AB,X_1,X_2')  
X_A=sp.Function('X_A')(z)  
```

Le flux $N_{Az}$ est une constante, donc l'équation 18-2.1 écrite comme $N_{Az} = -\frac {c D_{AB}}{1 - X_A} \frac {dX_A}{dz}$ peut être séparée et intégrée entre les deux limites du domaine:

```python
eq=-C*D_AB/(1-X_A)
N_AZ=sp.integrate(eq,(X_A,X_1,X_2))/sp.integrate(1,(z,z_1,z_2))
display(N_AZ)
```
