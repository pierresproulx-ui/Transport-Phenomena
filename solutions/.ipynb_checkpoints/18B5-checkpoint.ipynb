```python
#
# Pierre Proulx
#
#
# Préparation de l'affichage et des outils de calcul symbolique
#
import sympy as sp
from IPython.display import *
sp.init_printing(use_latex=True)
%matplotlib inline
```

```python
# Paramètres, variables et fonctions

z,L,k_2,C_A0,D_AB,S,delta_z=sp.symbols('z,L,k_2,C_A0,D_AB,S,delta_z')  
N_A=sp.Function('N_A')
C_A=sp.Function('C_A')(z)    
```

```python
#
# Le bilan est identique à la section 18-4 
#
eq=(S*N_A(z)-S*N_A(z+delta_z))/delta_z-k_2*C_A*S
display(eq)
```

```python
eq=sp.limit(eq,delta_z,0).doit()
display(eq)
```

```python
fick=-D_AB*sp.diff(C_A,z)
eq=eq.subs(N_A(z),fick)
display(eq)
```

```python
C_A=sp.dsolve(eq,C_A)
display(C_A)
C=C_A                    # je garde une copie de l'expression pour plus tard
```

```python
C_A=C_A.rhs
#
# Pose et solution des 2 équations de conditions aux limites pour C1 et C2
constantes=sp.solve(
           [sp.Eq(C_A.diff(z).subs(z,L    ),0) # les conditions 
           ,sp.Eq(C_A.subs(z,0),C_A0)   ]       # aux limites
           ,sp.symbols('C1,C2') )             # pour trouver C1 et C2
display(constantes)
```

```python
C_A1=C_A.subs(constantes)
display(C_A1.simplify())
```

On peut maintenant utiliser la solution pour calculer , par exemple, la quantité de la substance A qui réagit dans le réacteur. Elle est obtenue en multipliant le flux molaire en z=0 par la surface, donc:

$ W_A = S * N_A $ en z=0 donc $ W_A = S * (-D_{AB}) \frac {dC_A}{dz}_{z=0} $

```python
W_A=S*C_A.diff(z).subs(z,0)
display(W_A)
```

Le problème développé à la section 18.4 est identique au problème d'ailette traité à la section 10.7. 

Il ne serait pas beaucoup plus difficile de traiter ce problème différemment. Par exemple, au lieu de considérer la fond du réacteur comme imperméable, on pourrait considérer qu'il contient un catalyseur rapide. La condition en z=L deviendrait que la concentration est nulle

```python
#
# Reprenons C, la copie de C_A avant d'y substituer les conditions aux limites
#
C_A=C.rhs
display(C_A)
```

```python
#
# Pose et solution des 2 équations de conditions aux limites pour C1 et C2
constantes=sp.solve(
           [sp.Eq(C_A.subs(z,L),0)  
           ,sp.Eq(C_A.subs(z,0),C_A0)   ]      
           ,sp.symbols('C1,C2') )             
display(constantes)
```

```python
C_A2=C_A.subs(constantes)
display(C_A2.simplify())
```

Autre exemple, au lieu de considérer la fond du réacteur comme imperméable, on pourrait considérer qu'il contient un catalyseur réel. La condition en z=L deviendrait que la concentration est déterminée par la vitesse de réaction de surface donc: $ -D_{AB} \frac {dC_A}{dz} = k_s C_A $ en z=L

```python
#
# Reprenons encore C, la copie de C_A avant d'y substituer les conditions aux limites
#
C_A=C.rhs
display(C_A)
```

```python
#
# Pose et solution des 2 équations de conditions aux limites pour C1 et C2
k_s=sp.symbols('k_s')
constantes=sp.solve(
           [sp.Eq(-D_AB*C_A.diff(z).subs(z,L)-k_s*C_A.subs(z,L),0) 
           ,sp.Eq(C_A.subs(z,0),C_A0)   ]      
           ,sp.symbols('C1,C2') )             
display(constantes)
```

```python
C_A3=C_A.subs(constantes)
display(C_A3.simplify())
```
