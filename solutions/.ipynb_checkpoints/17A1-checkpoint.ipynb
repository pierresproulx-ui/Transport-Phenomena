### Problème 17A1  # solution utilisant thermo


```python
#A)
import thermo as th
# fonction Dab équation 17.2-1, utile aussi pour extrapoler
def Dab(A,B,p,T,MA,MB,polaire): #  cm2/sec
    a=[2.745e-4,3.64e-4]
    b=[1.823,2.334]
    TcA=th.critical.Tc(CASRN=A)
    PcA=th.critical.Pc(CASRN=A)/101325
    TcB=th.critical.Tc(CASRN=B)       #  pression en atm...
    PcB=th.critical.Pc(CASRN=B)/101325
    f1=(PcA*PcB)**(1/3)
    f2=(TcA*TcB)**(5/12)
    f3=(1/MA+1/MB)**(1/2)
    f=f1*f2*f3
    if polaire:
        return f*a[1]*(T/(TcA*TcB)**0.5)**b[1]/p   # polaire
    else:
        return f*a[0]*(T/(TcA*TcB)**0.5)**b[0]/p   # non-polaire
# fin de la fonction Dab
methane=th.Chemical('CH4')   # l'objet Chemical, ses fonctions dont
ethane=th.Chemical('C2H6')   # molar weight (.MW) 
T=293                        # le nom chimique standard (.CAS), etc...
P=1
Dab(methane.CAS,ethane.CAS,P,T,methane.MW,ethane.MW,polaire=False) 
```

#### c) Utilisons les valeurs critiques provenant de l'annexe de BSL Transport Phenomena


```python
# C)  
import math as math
def omegaDAB(T,epK):   # cette expression provient de l'annexe E
    T=T/epK
    A=1.06036/T**0.1561
    A+=0.193/math.exp(0.47635*T)
    A+=1.03587/math.exp(1.52996*T)
    A+=1.76474/math.exp(3.89411*T)
    return A
def DabCE(T,ep,s,MA,MB):
    return 0.0018583*(T**3*(1/MA+1/MB))**0.5/(p*s**2*omegaDAB(T,ep))
#
sA=3.78   # Annexe E
epA=154
sB=4.388
epB=232
#
sAB=0.5*(sA+sB)
epAB=(epA*epB)**0.5
T=293
p=1
DabCE(T,epAB,sAB,methane.MW,ethane.MW)
```
