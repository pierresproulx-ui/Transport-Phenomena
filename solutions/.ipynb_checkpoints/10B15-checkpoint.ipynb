### Problème 10B15



```python
#
# Pierre Proulx
#
# Préparation de l'affichage et des outils de calcul symbolique
#
import sympy as sp
from IPython.display import *
sp.init_printing(use_latex=True)
%matplotlib inline
# Paramètres, variables et fonctions

S_c,r_0,r_1,k_eff,T_0=sp.symbols('S_c,r_0,r_1,k_eff,T_0')        
r=sp.symbols('r',positive=True)
q=sp.Function('q')
T=sp.Function('T')(r) 

#
# Résultat du bilan 
#
eq=(sp.diff(q(r)*r,r)-r*S_c)
display('Equation différentielle obtenue du bilan, flux',eq)
eq=sp.dsolve(eq)
display(eq)
Fourier=-k_eff*T.diff(r)
# On substitue la loi de Fourier pour trouver le profil de température
eqT=eq.subs(q(r),Fourier)
T=sp.dsolve(eqT)
display('Profil de température général, avant application des conditions aux limites',T)
#limites
condition_1=sp.diff(T.rhs,r).subs(r_0,0)
condition_2=T.subs(r,r_0).rhs-T_0
display('Conditions aux limites',condition_1,condition_2)
constantes=sp.solve([condition_1,condition_2],sp.symbols('C1,C2'))
display('Constantes obtenues',constantes)
T=T.subs(constantes)
display('Profil de température final',T)
dico={'r_0':0.01,'r_1':0.011,'k_eff':1,'T_0':200,'S_c':10}
Tp=T.subs(dico)
display(Tp)
display(-k_eff*Tp.rhs.diff(r).evalf(3))
```
