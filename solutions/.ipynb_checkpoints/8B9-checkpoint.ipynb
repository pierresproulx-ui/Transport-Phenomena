## Exercice 8B9, on refait 7B9 pour un fluide d'Ostwald-de Waele

```python
#
# Initialiser sympy et l'affichage des équationa
#
import sympy as sp
from IPython.display import *
sp.init_printing(use_latex=True)
%matplotlib inline
H,r,L,rho,g,m,R,R_0,t=sp.symbols('H,r,L,rho,g,m,R,R_0,t')
n=sp.symbols('n',positive=True)
tau=sp.Function('tau_rz')(r)
v_z=sp.Function('v_z')(r)
h=sp.Function('h')(t)
tau_rz=m*(-sp.Derivative(v_z,r))**n # power-law
eq1=sp.Eq(tau_rz-(rho*g*(h+L)/L)*r,0) # bilan, résultat de 2.3  
v_z=sp.dsolve(eq1,v_z).rhs # v_zs est la solution, à la droite du signe =
C1=sp.symbols('C1')
constante=sp.solve((v_z.subs(r,R_0),0),C1)
v_z=v_z.subs(constante)
display(v_z.simplify())
w=2*sp.pi*rho*sp.integrate(v_z*r,(r,0,R))
w
```

```python
eq_vidage=sp.Eq(sp.pi*R**2*rho*sp.diff(h,t),w)
```

```python
eq_vidage.simplify().rhs
```

```python
eq_vidage=eq_vidage.rhs*((n+1)*(3*n+1))
eq_vidage.simplify()
```
