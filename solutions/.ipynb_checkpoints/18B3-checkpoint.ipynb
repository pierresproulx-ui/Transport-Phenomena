### Problème 18B3

```python
#
# Pierre Proulx
#
#
# Préparation de l'affichage et des outils de calcul symbolique
#
import sympy as sp
from IPython.display import *
sp.init_printing(use_latex=True)
%matplotlib inline
```

```python
# Paramètres, variables et fonctions
z,z_1,z_2,C,D_AB,X_1,X_2=sp.symbols('z,z_1,z_2,C,D_AB,X_1,X_2')  
X_A=sp.Function('X_A')(z)  
```

On utilise comme dans 18B1 mais en intégrant 18.2-1 entre $z_1$ et $z$ , un point quelque part entre $z_1$ et $z_2$ et z au lieu d'intégrer entre les limites  $z_1$ et $z_2$

```python
eq=-C*D_AB/(1-X_A)
eq1=sp.integrate(eq,(X_A,X_1,X_A))/sp.integrate(1,(z,z_1,z))
N_AZ=sp.Function('N_AZ')(z)
N_AZ=sp.Eq(N_AZ,eq1)
display(N_AZ)
```

Un peu de manipulations et on arrivera rapidement à l'équation 18B.3-1

```python
N_AZr=(N_AZ.rhs*((z-z_1)/(C*D_AB))).simplify()
N_AZl=(N_AZ.lhs*((z-z_1)/(C*D_AB))).simplify()
display(N_AZr)
display(N_AZl)
#
eq1=sp.exp(N_AZl).simplify()
eq2=sp.exp(N_AZr).simplify()
#
eq=sp.Eq(eq2,eq1) 
eq1=sp.exp(N_AZl).simplify()
eq2=sp.exp(N_AZr).simplify()
eq=sp.Eq(eq2,eq1)
display(eq)
```
