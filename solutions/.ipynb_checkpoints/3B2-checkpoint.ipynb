```python
# Exercice 3B2
#
import sympy as sp
from ipywidgets import *
from IPython.display import *
import fluids as fl
from fluids.units import *
sp.init_printing(use_latex=True)
x,y,H,DP,L,H,mu,rho=sp.symbols('x,y,H,DP,L,H,mu,rho')
v_z=sp.Function('V_z')(x,y)
v_zs=DP/(4*mu*L*H)*(y-H)*(3*x**2-y**2)
```

```python
eq=sp.Eq((sp.diff(v_zs,x,x)+sp.diff(v_zs,y,y))*mu+DP/L,0)
display(eq)
display(eq.simplify())
```

```python
debitMassique=2*rho*sp.integrate(v_zs,(x,0,y/3**(1/2)),(y,0,H))
debitMassique
```

```python
surface=H**2/3**(1/2)
vitesseMoyenne=debitMassique/rho/surface
vitesseMoyenne
```

```python
# La position sera forcément au centre, en x=0, donc
v_c=v_zs.subs(x,0)
positionMax=sp.solve(sp.Eq(sp.diff(v_c,y),0),y)
display(positionMax)
positionMax[1]
```

```python
# vitesse moyenne par rapport à la vitesse maximum
vitesseMoyenne/v_c.subs(y,positionMax[1])
```
