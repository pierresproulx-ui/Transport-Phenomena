```python
import thermo as th
### Exercice 9A10 avec thermo
frac=0.25
N,O,C=(1-frac)*0.8,(1-frac)*0.20,frac
Cl1=th.Mixture(['nitrogen', 'oxygen', 'chlorine'], zs=[N,O,C], T=297, P=101325)
print('25% chlore',Cl1.kg,N+O+C)
frac=0.5
N,O,C=(1-frac)*0.8,(1-frac)*0.2,frac
Cl2=th.Mixture(['nitrogen', 'oxygen', 'chlorine'], zs=[N,O,C], T=297, P=101325)
print('50% chlore',Cl2.kg)
frac=0.75
N,O,C=(1-frac)*0.8,(1-frac)*0.2,frac
Cl3=th.Mixture(['nitrogen', 'oxygen', 'chlorine'], zs=[N,O,C], T=297, P=101325)
print('75% chlore',Cl3.kg)
print('calculé avec thermo version ',th.__version__) 
```
