```python
import sympy as sp
from IPython.display import *
sp.init_printing(use_latex=True)
%matplotlib inline
import numpy as np
from ipywidgets import *
def f(Fin):
    rho,c_p,v_0,k_eff,S,z,F=sp.symbols('rho,c_p,v_0,k_eff,S,z,F')
    #F=rho*v_0*c_p/k_eff     # regrouper ce terme pour simplifier un peu la notation
    T_1=sp.Function('T_1')(z)
    T_2=sp.Function('T_2')(z)
    T_3=sp.Function('T_3')(z)
    eq1=sp.Eq(sp.diff(T_1,z,z)-F*sp.diff(T_1,z),0)
    eq2=sp.Eq(sp.diff(T_2,z,z)-F*sp.diff(T_2,z)+S*T_2,0)
    eq3=sp.Eq(sp.diff(T_3,z,z)-F*sp.diff(T_3,z),0)
    T_1=sp.dsolve(eq1,T_1).rhs
    T_2=sp.dsolve(eq2,T_2).rhs  # rhs pour avoir seulement a droite de l'égalité
    T_3=sp.dsolve(eq3,T_3).rhs
    T_2=T_2.subs(sp.symbols('C1'),sp.symbols('C3'))
    T_2=T_2.subs(sp.symbols('C2'),sp.symbols('C4'))
    T_3=T_3.subs(sp.symbols('C1'),sp.symbols('C5'))
    T_3=T_3.subs(sp.symbols('C2'),sp.symbols('C6'))
    q_1 =-k_eff*sp.diff(T_1,z)
    q_2 =-k_eff*sp.diff(T_2,z)
    q_3 =-k_eff*sp.diff(T_3,z)
    group=sp.sqrt(F*F-4*S)
    G=sp.symbols('G')
    T_1=T_1.subs(group,G)
    T_2=T_2.subs(group,G)
    T_3=T_3.subs(group,G)
    q_1=q_1.subs(group,G)
    q_2=q_2.subs(group,G)
    q_3=q_3.subs(group,G)
    T_0,z_1,z_2=sp.symbols('T_0,z_1,z_2')
    CL1=sp.Eq(T_1.subs(z,0)-T_0,0)
    CL2=sp.Eq(T_1.subs(z,z_1)-T_2.subs(z,z_1),0)
    CL3=sp.Eq(q_1.subs(z,z_1)-q_2.subs(z,z_1),0)
    CL4=sp.Eq(T_2.subs(z,z_2)-T_3.subs(z,z_2),0)
    CL5=sp.Eq(q_2.subs(z,z_2)-q_3.subs(z,z_2),0)
    CL6=sp.Eq(q_3.subs(z,sp.oo),0)               # cette condition force C6=0
    C6=0
    dico={'F':Fin,'c_p':1000,'z_1':1,'z_2':2,'G':3,'T_0':293,'C6':0,'G':1,'k_eff':1}
    CL1=CL1.subs(dico)
    CL2=CL2.subs(dico)
    CL3=CL3.subs(dico)
    CL4=CL4.subs(dico)
    CL5=CL5.subs(dico)
    constantes=sp.solve([CL1,CL2,CL3,CL4,CL5]
                        ,sp.symbols('C1,C2,C3,C4,C5'))
    T_1=T_1.subs(constantes)
    T_2=T_2.subs(constantes)
    T_3=T_3.subs(constantes)
    q_1=q_1.subs(constantes)
    q_2=q_2.subs(constantes)
    q_3=q_3.subs(constantes)
    #dico={'F':5,'c_p':1000,'z_1':1,'z_2':2,'G':3,'T_0':293,'C6':0,'G':1,'k_eff':1}
    T_1plot=T_1.subs(dico)
    T_2plot=T_2.subs(dico)
    T_3plot=T_3.subs(dico)
    import matplotlib.pyplot as plt
    plt.rcParams["figure.figsize"]=5,4
    sp.plot((T_1plot,(z,0,1)),(T_2plot,(z,1,2)),(T_3plot,(z,2,3)))
    plt.show()
g=interactive(f,Fin=(1,20,1))
display(g)
```
