### Problème 18B15 réacteur catalytique, 1 pore

```python
#
# Pierre Proulx
#
# Préparation de l'affichage et des outils de calcul symbolique
#
import sympy as sp
from IPython.display import *
sp.init_printing(use_latex=True)
%matplotlib inline
```

```python
# Paramètres, variables et fonctions
z,delta_z,rho,D_AB,k1,S,P,L=sp.symbols('z,delta_z rho D_AB k_1 S P L')
n_a=sp.Function('n_a')
omega_a=sp.Function('omega_a')(z)     
```

<img src="images/18B15.png" width=50% height=50%>

#### La partie a) est résolue en faisant un bilan, sachant que la réaction sur les parois fait disparaitre la substance a, le problème devient similaire à l'ailette ou la chaleur disparait sur la surface de l'ailette. La perte est une fonction f qui dépends de la concentration sur la paroi, $\omega_{a0}$ qui est fonction de z.
##### Bilan sur une section cylindrique  entre z et z+$\Delta$ z 
##### $ (S n_{az})_z -(S n_{az})_{z+\Delta z} + - P \Delta z f(\omega_{a0})=0 $
##### continuez...
##### la partie b vous vous en occupez
##### pour la partie c, la fonction $f(\omega_{a0})$ devient une fonction de z seulement (c-a-d que la concentration dans une section du pore est constante, on écrira donc $f(\omega_{a0})= k_1^{,,} \omega_a(z)$. En substituant ceci et la loi de Fick dans l'équation ci-haut on aura:

```python
eq=sp.Eq(D_AB*sp.diff(omega_a,z,z)-P/S*k1*omega_a,0)
display(eq)
```

   ## remarquez que cette équation est analogue à celle développée dans la section 10.7 pour les ailettes thermiques, et elle a les mêmes conditions aux limites.

```python
omega_a=sp.dsolve(eq,omega_a)
omega_a=omega_a.rhs
display('Solution générale',omega_a)
# Pose et solution des 2 équations de conditions aux limites pour C1 et C2
eq1=sp.Eq(omega_a.diff(z).subs(z,L),0)
eq2=sp.Eq(omega_a.subs(z,0),sp.symbols('omega_0'))
display('équations des conditions aux limites',eq1,eq2)
constantes=sp.solve([eq1,eq2],sp.symbols('C1,C2'))    # pour trouver C1 et C2
display('Valeurs des constantes après pose des conditions aux limites',constantes)
omega_a=omega_a.subs(constantes)
display('Profil de concentration après substitution des constantes', omega_a.simplify())
```

#### Le taux de réaction est calculé comme le taux de chaleur dissipé par une ailette est calculé dans la section 10.7. Le flux de réactif en z=0 multiplié par la surface est une méthode, et le taux de réaction à la paroi multiplié par la surface de la paroi est une autre méthode (retournez voir la section 10.7 et assurez-vous de bien comprendre ceci)

```python
taux=-D_AB*omega_a.diff(z).subs(z,0) # flux en z=0, kg / sec / m2
rate=taux*S                     # multiplié par la section donne le taux en kg / sec
display('Taux de réaction ',rate.simplify(),'kg/sec')
```

```python

#
# pour comparaison avec le livre, on choisira des valeurs arbitraires simples
#
dico={'P':1,'S':1,'k_1':1,'rho':1,'D_AB':1,'L':1,'omega_0':1,'z':1}
# ces valeurs sont pour poser N=1
omega=omega_a.subs(dico)
display(' comparaison de nos résultats avec ceux donnés dans Transport Phenomena pour N=1')
display(' par notre analyse',omega.evalf())
display('Transport phenomena',(sp.cosh(0)/sp.cosh(1)).evalf())
```
