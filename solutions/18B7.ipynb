### Problème 18B7, diffusion à partir d'une sphère suspendue dans l'air


```python
#
# Pierre Proulx
#
# Préparation de l'affichage et des outils de calcul symbolique
#
import sympy as sp
from IPython.display import *
sp.init_printing(use_latex=True)
```

```python
# Paramètres, variables et fonctions
r,D_AB,delta_r,r_1,r_2,c,X_B1,X_B2=sp.symbols('r,D_AB,delta_r,r_1,r_2,c,X_B1,X_B2')                                                            
X_A=sp.Function('X_A')(r)
N_Ar=sp.Function('N_Ar')
```

```python
#                                 bilan sur une coquille sphérique, pas de réaction
eq1=sp.diff(N_Ar(r)*r**2,r)
display(eq1)
eq2=sp.dsolve(eq1)
display(eq2)
```

```python
N_A1=sp.symbols('N_A1')
eq3=sp.Eq(eq2.rhs.subs(r,r_1)-N_A1,0)
C1=sp.symbols('C1')
constante=sp.solve(eq3,C1,dict=True)
display(constante)
constante=constante.pop()
eq2=eq2.subs(constante)
display(eq2)
fick=-c*D_AB/(1-X_A)*X_A.diff(r)
eq4=eq2.subs(N_Ar(r),fick)
display(eq4)
#               
#                On peut séparer l'équation 4 facilement à la main, remplacons 1-X_A par X_B
#
X_B=sp.symbols('X_B')
eq4l=-1/X_B
display(eq4l)
eq4r=eq4.rhs/(c*D_AB)
display(eq4r)
eq5=sp.Eq(sp.integrate(eq4l,(X_B,X_B1,X_B2)),0)         
display(eq5)
eq6=sp.Eq(sp.integrate(eq4r,(r,r_1,r_2)),0)
display(eq6)
#
#
Nfinal=sp.solve(eq5.lhs-eq6.lhs,N_A1)
display(Nfinal)
```
