```python
### Exercice 9A5 avec thermo
import thermo as th
from fluids.units import *
T=273.15+40
P=40*100000   # 1 bar = 100000 Pa , c'est encore une unité de confort, à éviter
eau=th.Chemical('water',T=T,P=P)
print("Conductivité de l'eau liquide à {:<4.0f} K, et {:<4.0f} Pa, {:<6.3} W/m-K".format(T,P,eau.kl))
klEau=eau.kl*u.W/u.K/u.m
print(klEau.to(u.BTU/u.hr/u.ft/u.degR))
```
