### Problème 10B1, on s'inspire de la section 2.3 mais le problème est beaucoup moins complexe, alors on a gardé essentiellement l'équation de T et l'équation de q

```python
#
# Pierre Proulx
#
# Préparation de l'affichage et des outils de calcul symbolique
#
import sympy as sp
from IPython.display import *
sp.init_printing(use_latex=True)
```

```python
# Paramètres, variables et fonctions
r,R,T_R,T_inf,delta_r,k=sp.symbols('r,R,T_R,T_inf,delta_r,k')                                                            
q=sp.Function('q')           # Notation de Transport Phenomena
T=sp.Function('T')(r) 
```

```python
#
# Résultat du bilan
#
eq=sp.diff(q(r)*r**2,r)
eq_T=eq.subs(q(r),-k*sp.diff(T,r))
T=sp.dsolve(eq_T,T)
display('Solution générale',T)
```

```python
# Conditions aux limites
condition_1=sp.Eq(T.rhs.subs(r,R),T_R)
condition_2=sp.Eq(T.rhs.subs(r,sp.oo),T_inf)
display('conditions aux limites',condition_1,condition_2)
constantes=sp.solve([condition_1,condition_2],sp.symbols('C1,C2'))
display('Les constantes C1 et C2',constantes)
T=T.subs(constantes)
display('Profil de température final après substitution des constantes',T)
q=-k*T.rhs.diff(r)
display(q)
h=sp.symbols('h')
eq3=sp.Eq(q-h*(T_R-T_inf),0)
display('Le flux de Fourier est égal au flux de Newton',eq3)
display('En résolvant cette dernière équation pour trouver h=')
eq4=sp.solve(eq3.subs(r,R),h)[0]
display(sp.Eq(h,eq4))
#
# Ce dernier résultat est exactement ce qui est demandé c-a-d puisque h=k/R, hD/k = 2. 
#
```
