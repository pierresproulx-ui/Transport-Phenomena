#### Exercice 1A4: puisque thermo permet facilement de calculer les viscosités de mélanges, il n'est pas plus difficle de tracer le graphe complet avec python.

```python
from matplotlib import rcParams
import matplotlib.pyplot as plt
%matplotlib inline
import thermo as th
import fluids
from fluids.units import *
import numpy as np
```

```python
melange=np.linspace(start=0,stop=1,num=10)
viscosite=np.zeros(len(melange))
for i,z in enumerate(melange):
    viscosite[i]=th.Mixture(['H2','CCl2F2'],zs=[1-z,z],T=298,P=101325).mu
    print(melange[i],viscosite[i])
```
