```python
# Exercice 1B1
#
import sympy as sp
from IPython.display import *
sp.init_printing(use_latex=True)
mu=sp.symbols('mu')
kappa=sp.symbols('kappa')
rho=sp.symbols('rho')
x,y,z,b=sp.symbols('x y z b')
#
# on fera seulement ici la partie d)
#
vx=-1/2*b*x
vy=-1/2*b*y
vz=b*z
#
# annexe B1
tauxx=-2*mu*sp.diff(vx,x)+(2/3*mu-kappa)*(sp.diff(vx,x)+sp.diff(vy,y)+sp.diff(vz,z))
tauyy=-2*mu*sp.diff(vy,y)+(2/3*mu-kappa)*(sp.diff(vx,x)+sp.diff(vy,y)+sp.diff(vz,z))
tauzz=-2*mu*sp.diff(vz,z)+(2/3*mu-kappa)*(sp.diff(vx,x)+sp.diff(vy,y)+sp.diff(vz,z))
tauxy=-mu*(sp.diff(vy,x)+sp.diff(vx,y))
tauyx=tauxy
tauxz=-mu*(sp.diff(vz,x)+sp.diff(vx,z))
tauzx=tauxz
tauyz=-mu*(sp.diff(vy,z)+sp.diff(vz,y))
tauzy=tauyz
rhovxvx=rho*vx*vx
rhovyvy=rho*vy*vy
rhovzvz=rho*vz*vz
rhovxvy=rho*vx*vy
rhovyvx=rhovxvy
rhovxvz=rho*vx*vz
rhovzvx=rhovxvz
rhovyvz=rho*vy*vz
rhovzvy=rhovyvz
print('Tau xx,yy,zz,xy,xz,yz')
display(tauxx,tauyy,tauzz,tauxy,tauxz,tauyz)
print('rhovv xx,yy,zz,xy,xz,yz')
display(rhovxvx,rhovyvy,rhovzvz,rhovxvy,rhovxvz,rhovyvz)
```
