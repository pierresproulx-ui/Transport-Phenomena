### Problème 2B7 avec modification, $P_0 \ne P_L$, par exemple une pression plus grande du côté droit pousse l'écoulement de la droite vers la gauche alors que la tige tire de la gauche vers la droite. Cet exercice est une modification mineure de l'exercice 2B7, on l'a solutionné en modifiant 2 lignes et en ajoutant une ligne.

### Observez bien dans les résultats comment la présence de la pression plus grande en z=L affecte l'écoulement. On pourrait, en jouant sur la pression $P_L$, obtenir un débit volumétrique nul.

<img src='images\Chap-2-Prob-2B7.png'>

```python
#
# Pierre Proulx
#
# Préparation de l'affichage et des outils de calcul symbolique
#
import sympy as sp
from IPython.display import *
sp.init_printing(use_latex=True)
%matplotlib inline
```

```python
# définitions variables, paramètres.
r,L,mu,R,rho,g,kappa,v_0=sp.symbols('r,L,mu,R,rho,g,kappa,v_0')
v_z=sp.Function('v_z')(r)
tau_rz=sp.Function('tau_rz')(r)
P0,PL,L=sp.symbols('P_0 P_L L')          ###########ajouté####################
```

```python
# Je reprends le bilan fait à la section 2.3, équation 2.3-10 sans le terme de gravité
eq1=r*tau_rz
eq2=eq1.diff(r)-(P0-PL)*r/L              ##########modifié#####################
#display(eq2)
```

```python
eq3=sp.dsolve(eq2)
#display(eq3)               
```

```python
eq4=eq3.subs(tau_rz,-mu*v_z.diff(r))     # substituer la loi de Newton pour le taux
#display(eq4)
```

```python
v_z=sp.dsolve(eq4)
#display('profil de vitesse, solution générale avant de trouver C1 et C2',v_z)
```

```python
v_z=v_z.rhs
condition_1=sp.Eq(v_z.subs(r,R),0)
condition_2=sp.Eq(v_z.subs(r,kappa*R)-v_0,0)
display('conditions aux limites',condition_1,condition_2)
constantes=sp.solve([condition_1,condition_2], sp.symbols('C1,C2'))
#display(constantes)
```

```python
v_z=v_z.subs(constantes)
#display('profil de vitesse après substitution des constantes', v_z)
```

```python
#
# Débit volumétrique
#
debit=sp.integrate(v_z*r,(r,kappa*R,R))
#display(debit.simplify())
```

```python
dic={'mu':0.001,'R':0.05,'kappa':0.5,'v_0':1,'P_0':0,'L':1} ###modifié###
v_zp=v_z.subs(dic)
kappa=kappa.subs(dic)
R=R.subs(dic)
import matplotlib.pyplot as plt          ### 
plt.rcParams['figure.figsize']=8,6     ### 
from ipywidgets import *
dic={'mu':0.001,'R':0.05,'kappa':0.5,'v_0':1,'P_0':0,'L':1} ###modifié###
def plotinter(DP):
    v_zp=((v_z.subs(dic))).subs(PL,DP)
    titre='Delta P ='+str(DP)
    sp.plot(v_zp,(r,kappa*R,R),title=titre);
interact(plotinter,DP=(-20,20,1))
```
