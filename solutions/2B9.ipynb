```python
#
# Préparation de l'affichage et des outils de calcul symbolique
#
import sympy as sp          
from IPython.display import *
sp.init_printing(use_latex=True)
%matplotlib inline
```

```python
# Paramètres, variables et fonctions
#
r,L,mu,R,P_0,P_L,rho=sp.symbols('r,L,mu,R,P_0,P_L,rho') # pas d'effet de gravité, seulement P
vz=sp.Function('v_z')(r)
C1=sp.symbols('C1')
```

>> On a déjà fait le bilan à la section 2.3, il vous est facile de vérifier que l'on peut maintenant partir de l'équation 2.3-15 puisque la condition limite au centre de la conduite est identique.

```python
eq2315=sp.Eq(vz.diff(r)+r*(P_0-P_L)/(2*mu*L),0)
display(eq2315)
```

```python
xi,xi0=sp.symbols('zeta,zeta_0')
vzg=sp.dsolve(eq2315,vz)
display(vzg)
```

```python
vzg=vzg.rhs
Pavg=sp.symbols('P')
Pavg=(P_0+P_L)/2
xi=xi0/Pavg
eq=sp.Eq(vzg.subs(r,R)+xi*vzg.diff(r),0)
display(eq)
```

```python
constante=sp.solve(eq,C1,dict=True) 
display(constante)          # dans ce cas, la solution est donnée dans une liste de 1 élément, sortons-le avec constante[0]
display('voici la constante C1, qui était dans une liste, je prends le premier élément [0]',constante[0])
vz=vzg.subs(constante[0])    # 
display(vz.simplify())
```

```python
w=2*sp.pi*rho*sp.integrate(r*vz,(r,0,R))
display(w.simplify())
```
