```python
# Exercice 3A1
#
import sympy as sp
from ipywidgets import *
from IPython.display import *
import fluids as fl
from fluids.units import *
sp.init_printing(use_latex=True)
mu=200*u.cP
rho=50*u.lb/u.ft**3
L=2*u.inch
Omega=200*u.min**(-1)
deltaR=0.002*u.inch
R=1*u.inch
```

```python
V=2*3.1416*R*Omega
gradient=V/deltaR
surface=L*2*3.1416*R
force=mu*gradient*surface
Power=force*V
```

```python
print('Puissance {:7.4f} ou {:7.4f}'.format(Power.to(u.hp),Power.to(u.kW)))
```
