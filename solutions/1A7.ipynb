```python
# Exercice 1A7
#
import thermo as th
```

```python
P=101325
T=273.2
k=1.38e-23
O2=th.Chemical('O2',T=273.2)
n=P/(k*T)                      # molecules / m3, c'est la loi des gaz parfaits 
d=O2.molecular_diameter/1e10   # donné en Angstrom par thermo, convertir en mètres
print(d)
N=6.023e26                     # unités S.I. pas des g-moles, mais des kg-moles.
MFP=1/(1.41*3.1416*d**2*n)     # mètres
VM=(8*k*T/3.1416/(32/N))**0.5  # mètres / sec
print (' vitesse moléculaire {:<7.1f} m/sec  et libre parcours moyen  {:<12.3e} m'.
      format(VM,MFP))
```
