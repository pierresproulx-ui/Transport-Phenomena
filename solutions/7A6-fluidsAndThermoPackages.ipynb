```python
import thermo as th
from math import *
from fluids.units import *
from thermo.units import Stream 
from scipy.optimize import fsolve, root
```

```python
#
# Thermo
#
T=68*u.degF
T.ito_base_units()
eau=th.Chemical('water',T=T.magnitude)
rho=(eau.rho)*u.kg/u.m**3
mu=(eau.mu)*u.kg/u.m/u.sec
#
# fluids
#
NPS, Di, Do, t = nearest_pipe(Di=5*u.inch)
```

```python
#
# pint (fluids.units)
#
L=12*u.ft+11*u.ft+14*u.ft
H=12*u.ft+14*u.ft+12*u.ft
#
H.ito_base_units()
L.ito_base_units()
Di.ito_base_units()
A=pi*Di**2/4
#
```

```python
# La différence de pression exercée sur le fluide (rho g H) doit être égale à la friction totale
# On l'écrit donc avec fluids cette égalité et on utilise fsolve qui solutionne itérativement.
#
def calcule_v(v):
    v=v*u.m/u.sec
    g=9.81*u.m/u.sec**2
    Re=rho*v*Di/mu
    fd = friction_factor(Re=Re)
    K_D = K_from_f(fd=fd, L=L, D=Di)   # pertes associées à la friction dans la conduite
    K_D+=2*bend_rounded(Di=Di,angle=90*u.degrees,fd=fd,bend_diameters=1.5) # dues aux coudes
    K_D+=entrance_sharp() # due à l'entrée dans la conduite
    K_D+=exit_normal()    # due à la sortie
    dP1 = rho*H*g
    dP2 = dP_from_K(K=K_D, rho=rho, V=v)
    return dP2-dP1
```

```python
vPourDemarrerLeCalcul=1*u.m/u.sec
v=(fsolve(calcule_v,vPourDemarrerLeCalcul))*u.m/u.sec
print(v)
print(v*A)
```
