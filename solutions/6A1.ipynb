```python
#
# Pierre Proulx
#
# Préparation de l'affichage et des outils de calcul symbolique
#
import sympy as sp
from IPython.display import *
sp.init_printing(use_latex=True)
# Paramètres, variables et fonctions
rho,L,D,mu,Q,dP,v_z=sp.symbols('rho,L,D,mu,Q,dP,v_z')                                                            
```

```python
f=1/4*(D/L)*(dP/(1/2*rho*v_z**2))    # equation définissant le facteur f
Re=4*Q*rho/(sp.pi*mu*D)              # Reynolds avec le débit volumique au lieu de la vitesse                     
f_T=0.0791/Re**0.25                  # si Re > 2100
# Dictionnaire contenant les valeurs des paramètres
dico={'rho':1000,'D':0.25,'mu':0.001,'L':1234,'Q':1.97,'L':1234+4*32*0.25+2*15*0.25}
#
# voir exemple 6.2-1
#
dP=2/D*rho*(4*Q/(sp.pi*D**2))**2*f_T*L
display('Reynolds = ',Re.subs(dico).evalf(2))
display('delta P (MPa) évalué avec f de Blasius, sous-estime la perte de pression pour Re > 100,000',dP.subs(dico).evalf(2)*1e-6)
display(' meilleure évaluation en lisant sur le graphe avec f ~ 0.02')
f_T=0.002
dP=2/D*rho*(4*Q/(sp.pi*D**2))**2*f_T*L
display(dP.subs(dico).evalf(2)*1e-6)
```

# Avec le package fluids

```python
import thermo as th
from math import *
from fluids.units import *
from thermo.units import Stream ## attention, il faut les versions récentes de fluids et thermo (0.1.68 et 0.1.39)
```

```python
Q = 1.97*u.m**3/u.s
T = (20+273)*u.degK
eau=th.Chemical('water',T=T.magnitude)
```

```python
Di=0.25*u.m
r_d=0.65  # pour les coudes...
L=(1234)*u.m
v = (Q/(pi*Di**2/4))
rho=eau.rho*u.kg/u.m**3
mu=eau.mu*u.kg/u.m/u.sec
Re = Reynolds(D=Di, rho=rho, mu=mu, V=v)
print(Re)
```

```python
fd=friction_factor(Re=Re)
K = K_from_f(fd=fd, L=L, D=Di)
K+=4 * bend_rounded(Di, angle=90 * u.degrees, fd=fd, bend_diameters=0.75)
K+=2 * bend_rounded(Di, angle=45 * u.degrees, fd=fd, bend_diameters=0.75)
dP = dP_from_K(K=K, rho=rho, V=v)
print('Reynolds ',Re)
print('Pression ',dP.to(u.psi),'\nPression  ',dP.to(u.Pa)/1e6,' MPa ')
```
