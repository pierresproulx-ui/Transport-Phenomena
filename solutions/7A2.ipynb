```python
Di=4*u.inch
rho=62.4*u.lb/u.ft**3
A1=pi*Di**2/4
v=2.3*u.ft/u.sec
L=300*u.ft
L.ito_base_units()
rho.ito_base_units()
v.ito_base_units()
Di.ito_base_units()
Re=7.11e4
mu=1*u.cp
mu.ito_base_units()
fd = friction_factor(Re=Re)
```

<center>
$
  \begin {split} \sum_{entrée} \bigg ( \frac {1}{2} v_1^2 + g h_1 + \frac {p_1}{\rho_1} \bigg)w-\sum_{sortie} \bigg ( \frac {1}{2} v_2^2 + g h_2 + \frac {p_2}{\rho_2} \bigg )w=\\-W_m +E_c+ w \frac {V^2}{2}  \big [ \sum_{i-conduites} {K_{f-sur-L}}_i +  \sum_{j-perturbations} K_j   \big ]
  \end {split}
  (16)$

<center>
$
  \boxed{
   w \bigg (  \frac {p_1}{\rho} \bigg)- w \bigg (  \frac {p_2}{\rho} \bigg )= -W_m +  w \frac {V^2}{2}  \big [ K_l +K_{entrée} +K_{sortie} \big ] 
       }
  $
    
    ou
    
</center>

<center>
$
  \boxed {
  ({p_1}- {p_2})- \rho \frac {V^2}{2} \big[ K_l +K_{entrée} +K_{sortie} \big] =
  -\frac {\rho}{w} W_m
          }
$
    
      En multipliant par -1 et en notant que le deuxième terme est celui calculé comme dP par fluids
</center>
 
<center>   
$
  \boxed{(p_2-p_1) + dP_{from_K} =  \frac {\rho}{w} W_m }
  $
    
    finalement
</center>

<center>   
$
  \boxed{\frac {w}{\rho} dP_{total} =   W_m }
  $
</center>

```python
K_D  = K_from_f(fd=fd, L=L, D=Di)           # la conduite
K_D +=contraction_sharp(Di1=Di*100,Di2=Di) # L'entrée
K_D +=diffuser_sharp(Di1=Di,Di2=Di*100)  # La sortie
dP = dP_from_K(K=K_D, rho=rho, V=v)      # dP total dans le système dû à la friction
P2mP1=3*u.atm               # dP de la différence entre les deux pressions dans les 2 réservoirs
P2mP1.ito_base_units()
dP += dP+P2mP1
w=rho*v*A1
puissance=dP*w/rho   #   À partir de (16 du notebook chapitre 7)
puissance.to(u.kW)
```
