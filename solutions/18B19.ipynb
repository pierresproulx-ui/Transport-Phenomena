```python
#
# Pierre Proulx
#
#
# Préparation de l'affichage et des outils de calcul symbolique
#
import sympy as sp
from IPython.display import *
sp.init_printing(use_latex=True)
%matplotlib inline
```

```python
N,xi,xi_0=sp.symbols('N,xi,xi_0',positive=True)
chi=sp.Function('chi')(xi)
eq=sp.Eq(sp.diff(xi**2*(sp.diff(chi,xi)),xi)-N*xi**2,0)
display(eq)
```

```python
chi=sp.dsolve(eq,chi)
display(chi)
```

```python
chi=chi.rhs
eq1=sp.Eq(chi.subs(xi,1),1)
eq2=sp.Eq(chi.diff(xi).subs(xi,xi_0),0)
constante=sp.solve([eq1,eq2], sp.symbols('C1,C2'))
display(constante)
```

```python
chi=sp.Eq(sp.simplify(-3/N*chi.subs(constante)),0) # pour mettre sous la même forme que Bird
display(chi)
```

```python
#chi=chi.lhs
xi0=sp.solve(chi.subs(xi,xi_0),xi_0)
#display(xi0)  # 2 racines, 1 racine complexe conjuguée (donc double) et 1 racine réelle
x=xi0.pop()   # la racine réelle
display(x.ratsimp())
```

```python
import matplotlib.pyplot as plt
plt.rcParams['figure.figsize']=10,8
sp.plot(x,(N,1,10))    # quand N=6, on commence à avoir une zone anoxie
```
