```python
#
# Pierre Proulx
#
# Préparation de l'affichage et des outils de calcul symbolique
#
import sympy as sp
from IPython.display import *
import thermo as th
sp.init_printing(use_latex=True)
# Paramètres, variables et fonctions
rho,L,D,mu,Q,dP,v_z=sp.symbols('rho,L,D,mu,Q,dP,v_z')                                                            
```

```python
f=1/4*(D/L)*(dP/(1/2*rho*v_z**2))    # equation définissant le facteur f
v_z=Q/(sp.pi*D**2/4)
Re=4*Q*rho/(sp.pi*mu*D)              # Reynolds avec le débit volumique au lieu de la vitesse                     
f_T=0.0791/Re**0.25                      # si Re > 2100
# Dictionnaire contenant les valeurs des paramètres
# convertis en SI
LSI=(30+50+15)*0.3048+2*15*3.068*0.0254   # ( pieds en mètres 0.3048 et pouces en mètres 0.0254)
densite=th.Chemical('water',273+20,101325).rhol
viscosite=th.Chemical('water',273+20,101325).mul
debit=18*0.00378541/60   # m3/sec  ( conversion gallons en m^3 et en secondes)
dico={'rho':densite,'D':3.068*0.0254,'mu':viscosite,'L':LSI,'Q':debit}
#
angle=45*sp.pi/180 # (radians...)
dP=(4/D)*(1/2*rho*v_z**2)*f_T+ 9.81*rho*50*0.3048*sp.sin(angle)
display('Reynolds = ',Re.subs(dico).evalf())
dP0=dP.subs(dico).evalf()*14.7/101325
display('delta P (PSI) évalué avec f de Blasius',dP0)
display(' seulement la hauteur corresond à')
display((9.81*rho.subs(dico)*50*0.3048*sp.sin(angle)).evalf()*14.7/101325)
dP1=((4/D)*(1/2*rho*v_z**2)*f_T).subs(dico).evalf()*14.7/101325
display('donc la friction seulement ne donne que (en PSI)',dP1)
display(' soit en pourcentage',dP1/dP0*100)
```

```python
import thermo as th
from math import *
from fluids.units import *
from thermo.units import Stream ## attention, il faut les versions récentes de fluids et thermo (0.1.68 et 0.1.39)
```

```python
L=(15+50+30)*u.ft
H=(50*cos(45*u.degree))*u.ft    # conversion degrés-radians...
Q = 18*u.gal/u.min
T = 68*u.degF
eau=th.Chemical('water',T=T.to(u.degK).magnitude)  # seulement en unités SI ??
NPS, Di, Do, t = nearest_pipe(Di=3*u.inch)   # aller chercher le vrai tube le plus près de ceux existant dans les catalogues
v = (Q/(pi*Di**2/4))
rho=eau.rho*u.kg/u.m**3
mu=eau.mu*u.kg/u.m/u.sec
Re = Reynolds(D=Di, rho=rho, mu=mu, V=v)
print(Re)
```

```python
fd=friction_factor(Re=Re)
K = K_from_f(fd=fd, L=L, D=Di)
K+=2 * bend_rounded(Di, angle=45 * u.degrees, fd=fd, bend_diameters=0.65)
dP = dP_from_K(K=K, rho=rho, V=v)
dP1=dP
print('première partie, seulement la friction',dP)
dP+=rho*u.gravity*H
print('Maintenant en considérant la hauteur',dP)
print('Maintenant en considérant la hauteur',dP.to(u.psi))
print((dP-dP1)/(dP)*100,'% requis pour monter le fluide')
```
