```python
### Exercice 9A11, partie a sable mouillé
import numpy as np
k=np.array([1.42e-3, 20.4e-3 , 7e-3])
phi=np.array([0.427, 0.510 , 0.063])
alpha=np.array([0., 0. , 0.])
g=np.array([1/3,1/3,1/3])
num=0
den=0
for i in range(0,3):
    alpha[i]=alpha[i]+1/(1+(k[i]/k[0]-1)*g[i])
for i in range(0,3):
    num=num+alpha[i]*k[i]*phi[i]    
    den=den+alpha[i]*phi[i]
print(num/den)
print(alpha)
print(phi)
print(k)
```
