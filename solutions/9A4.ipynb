```python
### Exercice 9A4 avec thermo
import thermo as th
CO2=th.Chemical('CO2',T=300,P=101325)
H2=th.Chemical('H2',T=300,P=101325)
print('CO2 ',CO2.kg,' H2 ',H2.kg)
melange=th.Mixture(['CO2', 'H2'], zs=[.2, .8], T=300, P=101325)
print('Mélange CO2 80% et H2 20% ',melange.kg,' W/m-K')
```
