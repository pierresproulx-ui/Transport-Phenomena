```python
#
# Pierre Proulx
#
# Préparation de l'affichage et des outils de calcul symbolique
#
import sympy as sp
from IPython.display import *
import thermo as th
sp.init_printing(use_latex=True)
# Paramètres, variables et fonctions
rho,D,mu,v_z=sp.symbols('rho,D,mu,v_z')                                                            
```

```python
Re=rho*v_z*D/mu              #                      
f_T=0.0791/Re**0.25                      # si Re > 2100  et < 40000
erreur=f_T.subs(v_z,0.96)/f_T.subs(v_z,1)
display('erreur en %',erreur.simplify()*100-100)
```
