#### Bilan de quantité de mouvement
L'équation de bilan de quantité de mouvement (momentum) est celle qu'il nous faut pour trouver la force qui est exercée sur le coude, en prenant les points 1 et 2 à l'entrée et la sortie de la conduite (plan 1 pour l'entrée et plan 2 pour la sortie):
<center>
$\begin{equation*}
  \boxed{ \sum_{entrée} ( v_1w_1 + p_1 S_1) \vec{u_1}-\sum_{sortie}( v_2w_2 + p_2 S_2) \vec{u_2} + m_{tot} \vec g=\vec F_{fluide \rightarrow surface}} \end{equation*}$ (1)
</center>

On effectue un bilan de momentum dans la direction x seulement,c'est la force horizontale qui est recherchée.

Définissons la direction positive de l'axe x vers la droite. Dans ce cas, 
Au plan 1, l'écoulement est vers la droite. Le terme $\vec{u_1}$ peut se définir comme le produit scalaire du vecteur unitaire vitesse au point 1 par le vecteur unitaire $\vec \delta_x$. Ainsi à la surface 1, $\vec{u_1}$ est positif donc le produit scalaire $\vec {\delta_x} \cdot \vec {u_1} = 1$. Au point 2, la vitesse $\vec{u_2}$ est négative donc: $\vec {\delta_x} \cdot \vec {u_2} = -1$. 

Le dernier terme à gauche  de l'équation 1 est traité de la même façon:
$\vec {\delta_x} \cdot \vec g = 0$

L'équation 1 devient:
<center>
$\begin{equation*}
  \boxed{  ( v_1w_1 + p_1 S_1) +( v_2w_2 + p_2 S_2) ={F_{fluide \rightarrow surface}}_x}\end{equation*}$ (2)
</center>
Puisque les grandeurs des vitesses, les surfaces et les débits sont égaux, on a ainsi:
<center>
$\begin{equation*}
  \boxed{  2vw +(p_1 + p_2)S ={F_{fluide \rightarrow surface}}_x}
  \end{equation*}$ (3)
</center>

On peut maintenant utiliser **fluids** pour faire les calculs et prendre les unités en charge, ce qui n'est pas négligeable comme avantage car les unités données sont en **CGS** et en unités **impériales**.

```python
import thermo as th
from math import *
from fluids.units import *
from thermo.units import Stream 
#
# Calculs
#
rho=62.4*u.lb/u.ft**3   # livres par pieds cubes
mu=1*u.cP               # centipoises!!! mais elle ne sert pas dans ce problème...
p1=21*u.lbf/u.inch**2
p2=19*u.lbf/u.inch**2  # pression en livres-force par pouces carrés
q=3*u.feet**3/u.sec   # débit volumique en pieds cubes par secondes
w=q*rho               #  débit massique en livres par seconde  
S=(4*u.inch)**2*pi/4  # surface en pouces carrés
v=q/S                 # vitesse en unités mixtes
print('Vitesse ',v)
F=2*v*w+(p1+p2)*S     # force en unités complètement mélangées
print('Force ',F,' quelles belles unités, vraiment un beau système le système anglais.... bravo!')
print("J'utilise fluids pour mettre ces unités en ordre {:<6.0f} {:<6.0f} ".format(F.to(u.lbf),F.to(u.N)))
```
