```python
# Exercice 3B3
#
import sympy as sp
from ipywidgets import *
from IPython.display import *
import fluids as fl
from fluids.units import *
sp.init_printing(use_latex=True)
x,y,DP,L,B,mu,rho=sp.symbols('x,y,DP,L,B,mu,rho')
v_z=sp.Function('V_z')(x,y)
v_zs=DP*B**2/(4*mu*L)*(1-x**2/B**2)*(1-y**2/B**2)
```

```python
eq=sp.Eq((sp.diff(v_zs,x,x)+sp.diff(v_zs,y,y))*mu+DP/L,0)
display(eq)
display(eq.simplify())
```

```python
debitMassique=rho*sp.integrate(v_zs,(x,-B,B),(y,-B,B))
compareDebitsMassique=sp.Eq(debitMassique-0.563*DP*B**4*rho/(mu*L),0)
display(compareDebitsMassique.evalf().simplify())
```
