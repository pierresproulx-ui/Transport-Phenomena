Pas de terme de terme de compression ni de pompe, les termes de friction sont négligeables comparés à la pression et la hauteur, il ne reste que l'équation de Bernouilli, la partie de gauche de 7.6-1, équation D
<center>
$\begin{equation}
  \boxed{ \sum_{entrée} \bigg ( \frac {1}{2} v_1^2 + g h_1 + \frac {p_1}{\rho_1} \bigg)w_1-\sum_{sortie} \bigg ( \frac {1}{2} v_2^2 + g h_2 + \frac {p_2}{\rho_2} \bigg )w_2=-W_m +E_c+ \sum_{i-cond} (w_i 2 \frac  {L}{D} v^2 f)_i + \sum_{j-pert} (w_i \frac  {1}{2} v^2 e_v)_j}
  \end{equation} $(1)
</center>
devient, le débit étant le même aux deux points 1 et 2 (le point 1 est au haut du réacteur et 2 juste à la sortie)
<center>
$\begin{equation}
  \boxed{  \bigg ( \frac {1}{2} v_1^2 + g h_1 + \frac {p_1}{\rho_1} \bigg)-\bigg ( \frac {1}{2} v_2^2 + g h_2 + \frac {p_2}{\rho_2} \bigg )=0}\end{equation} $(2)
</center>

La pression est atmosphérique au point 2 (0), la hauteur de référence (0) est le point 2 et la vitesse est presque 0 au point 1:
<center>
$\begin{equation}
  \boxed{  \bigg ( g h_1 + \frac {p_1}{\rho_1} \bigg)-\bigg ( \frac {1}{2} v_2^2  \bigg )=0}\end{equation} $(3)
</center>
On peut maintenant calculer, il n'y a qu'une seule inconnue. On utilisera des données SI:

```python
import sympy as sp
#  lbs/pi3 en Kg/m3=16.0185
v_2=sp.symbols('v_2')
rho=65*16.0185
h1=20*0.3048
g=9.81
p1=100*101325/14.7  
g=9.81
R=8/2*0.0254
Bernouilli=g*h1 + p1/rho -1/2*v_2**2
v_2=sp.solve(Bernouilli,v_2)
display('Vitesse au point 2',v_2[1])     # puisque v est au carré, il y a deux racines, on prends la positive
w=v_2[1]*sp.pi*R**2*rho
display('débit massique en kg/sec au point 2',w.evalf())
```

### Partie b) Quantité de mouvement selon z maintenant

les points 1 et 2 maintenant (entrée et sortie) sont la sortie de la buse et la surface de la plaque.
ainsi le bilan général:

<center>
$\begin{equation*}
  \boxed{ \sum_{entrée} ( v_1w_1 + p_1 S_1) \vec{u_1}-\sum_{sortie}( v_2w_2 + p_2 S_2) \vec{u_2} + m_{tot} \vec g=\vec F_{fluide \rightarrow surface}}
  \end{equation*}$ (4)
</center>

deviendra (n'oubliez pas que le point 1 maintenant est le point 2 de la partie a):

<center>
$\begin{equation*}
  \boxed{ \sum_{entrée} ( v_1w +0) \vec{u_1}-\sum_{sortie}( 0 + 0) \vec{u_2} + m_{tot} \vec g= \vec F_{fluide \rightarrow surface}}
  \end{equation*}$ (5)
</center>

Le dernier terme de la partie de gauche représente l'effet de l'accélération gravitationnelle sur la colonne de liquide de 2 pouces de hauteur.
<center>

$\begin{equation*}
  \boxed{ \sum_{entrée} ( v_1w +0) \vec{u_1}-\sum_{sortie}( 0 + 0) \vec{u_2} + \pi R^2 (h_1-h_2) \rho \vec g=\vec F_{fluide \rightarrow surface}}
  \end{equation*}$ (6)
</center>

On peut maintenant solutionner

```python
dH=2*0.0254   # hauteur entre la plaque et la sortie
R=8/2*0.0254
Force=sp.symbols('Force')
mg=sp.pi*R**2*dH*rho*g
v_1=v_2[1]               # juste pour mettre la vitesse v_2 de la section précédente à v_1 ici
Momentum=v_1*w+mg-Force
Force=sp.solve(Momentum,Force)
display('Force en Newtons', Force[0]) 
```
