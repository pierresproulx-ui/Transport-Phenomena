#### Section 6.3, calcul de la vitesse terminale de chute libre d'une particule

Dans cet exemple on utilise une technique numérique comme dans l'exemple 6.2-2. On utilise toujours la solution des deux équations de f en pour trouver la vitesse avec la fonction fsolve de scipy. On a obtenu par un bilan de forces f pour une sphère: équation 6.1-7 qui donne f en fonction de $ v_{\infty}$: $$f=\frac {4}{3} \frac {gD}{v_{\infty}^2} \frac {\rho_{sph}-\rho}{\rho}$$ et nous avons aussi l'équation 6.3-16 qui donne f en fonction du nombre de Reynolds, donc de $ v_{\infty}$:
$$f=\bigg(\sqrt{ \frac{24}{Re} }+0.5407 \bigg)^2$$. Ces deux équations, pour le calcul de la vitesse terminale de chute libre d'une particule, contiennent deux inconnues: f et $v_{\infty}$

#### Solution, éliminer f en posant 6.1-7 = 6.3-16 et solutionner pour trouver $v_{\infty}$:
### $ \frac {4}{3} \frac {gD}{v_{\infty}^2} \frac {\rho_{sph}-\rho}{\rho} -\bigg(\sqrt{ \frac{24}{Re} }+0.5407 \bigg)^2 =0$ pour  $v_{\infty}$

```python
#
# Pierre Proulx
#
# Calcul utilisant numpy et scipy quand la solution analytique n'existe pas
# sympy ne peut  solutionner alors on utilise du calcul numérique: scipy
# Les techniques de recherche utilisées par scipy sont celles que vous verrez
# dans le cours GCB245
```

```python
import numpy as np
import math
import thermo as th
from scipy.optimize import fsolve, root # C'est le solver de scipy, un peu comme dans 'excel'
#
# définir la fonction dont on cherche les zéros
#
def Re(vinf):
    return rho*vinf*D/vis
def f(vinf):
    f1=((24/Re(vinf))**0.5+.5407)**2 # équation 6.3-16 pour f
    f2=4./3.*g*D/(vinf**2)*(rhop-rho)/rho         # équation 6.1-7 pour f
    return f1-f2    # les deux doivent être égales, donc f1-f2=0
```

```python
T,P=300,101325
air=th.Mixture(('nitrogen','oxygen'),zs=(0.8,0.2))
rho,vis=air.rhog,air.mug
g,D,rhop=9.81,0.003,2600    # valeurs en S.I, 0.003 m (3 mm) et 2600 kg/m3
vinf=fsolve(f,1e-10)   # fonction de recherche de zéros de scipy.optimize, valeur 'guess' de 1e-10
print (vinf,' Le nombre de Reynolds est ',Re(vinf))
```

### Et si on utilisait fluids de Caleb Bell?


```python
import fluids
vt=fluids.v_terminal(D=D, rhop=rhop, rho=rho, mu=vis)
Re=fluids.Reynolds(V=vt,rho=rho,mu=vis,D=D)
Cd=fluids.drag_sphere(Re)
print('Vitesse terminale {:<6.2f} m/s , Reynolds {:<6.0f}, Coefficient de trainée {:<6.4f}'.format(vt,Re,Cd))
print("Vous comprendrez que c'est tout ce dont on a besoin à partir de maintenant")
print("Vous auriez grand intérêt à vous souvenir des outils de fluids pour le cours d'opérations unitaires l'an prochain")
```
