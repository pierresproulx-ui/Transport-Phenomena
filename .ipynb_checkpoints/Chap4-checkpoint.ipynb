```python
import sympy as sp
from IPython.display import *
sp.init_printing(use_latex=True)
%matplotlib inline

y,t,nu=sp.symbols('y,t,nu')
                                                  
vz=sp.Function('v_z')(y,t)  
eq0=sp.Eq(sp.diff(vz,t),nu*sp.diff(vz,y,y))
display(eq0)
```

```python
eta1=sp.symbols('eta_1')
eta1=y/sp.sqrt(4*nu*t)
display(eta1)
eta=sp.symbols('eta')
phi=sp.Function('phi')(eta)
#
eq1=sp.Eq(sp.diff(phi,eta)*sp.diff(eta1,t)-nu*sp.diff(phi,eta,eta)*sp.diff(eta1,y)**2,0)
display(eq1)
```

```python
diviseur=-4*sp.sqrt(nu*t)
eq2=sp.Eq(eq1.lhs/diviseur,eq1.rhs/diviseur)
display(eq2.simplify())
#
#   Simplification faire à la main, avec la définition de eta 
#
eq3=sp.Eq(sp.diff(phi,eta,eta)+2*eta*sp.diff(phi,eta),0)
display(eq3)
```

```python
phi=sp.dsolve(eq3,phi)
display(phi)
```

```python
C1,C2=sp.symbols('C1 C2')
cl1=sp.Eq(phi.rhs.subs(eta,0),1)  
                            # condition à la surface de la patch 1
cl2=sp.Eq(phi.rhs.subs(eta,sp.oo),0)
                            # condition au loin, 0
constantes=sp.solve((cl1,cl2),sp.symbols('C1,C2'))
phi=phi.subs(constantes)
display(phi)
sp.plot(phi.rhs,(eta,0,2));
```
