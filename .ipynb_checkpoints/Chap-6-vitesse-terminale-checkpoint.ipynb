#### Section 6.3, calcul de la vitesse terminale de chute libre d'une particule

Dans cet exemple on utilise une technique numérique comme dans l'exemple 6.2-2. On utilise toujours la solution des deux équations de f en pour trouver la vitesse avec la fonction fsolve de scipy. On a obtenu par un bilan de forces f pour une sphère: équation 6.1-7 qui donne f en fonction de $ v_{\infty}$: $$f=\frac {4}{3} \frac {gD}{v_{\infty}^2} \frac {\rho_{sph}-\rho}{\rho}$$ et nous avons aussi l'équation 6.3-16 qui donne f en fonction du nombre de Reynolds, donc de $ v_{\infty}$:
$$f=\bigg(\sqrt{ \frac{24}{Re} }+0.5407 \bigg)^2$$. Ces deux équations, pour le calcul de la vitesse terminale de chute libre d'une particule, contiennent deux inconnues: f et $v_{\infty}$

### *Solution*

#### Il faut trouver la solution à ces deux équations à 2 inconnues, mais elles sont non-linéaires si le nombre de Reynolds n'est pas très petit (dans ce cas on a simplement f=24/Re). On ne peut donc pas solutionner facileemnt en isolant une des deux variables, il faut utikliser des méthodes de recherches numériques. Dans *fluids* de Caleb Bell, la solution numérique est faite:

### Pour une particule de 2 mm de diamètre et de densité 1000 $kg/m^3$ tombant dans l'air à -2 degrés C et pression atmosphérique

```python
import fluids as fl
import thermo as th
air=th.Mixture('air')
air.T=273.15-2
air.P=101325
D=0.002
rhop=1000
vt=fl.v_terminal(D=D, rhop=rhop, rho=air.rho, mu=air.mu)
Re=fl.Reynolds(V=vt,rho=air.rho,mu=air.mu,D=D)
f=fl.drag_sphere(Re)
print('Re={:5.0f} v={:5.2f} f={:6.4f}'.format(Re,vt,f))
```

### Essayons de voir si la méthode de calcul de Caleb Bell suit bien le graphe de la section 6.3

```python
from ipywidgets import *
import numpy as np
Re=[0.1,1,10,20,50,100,200,500,1000,2000,5000,10000]
f=np.zeros(len(Re))
for i in range(len(Re)):
    f[i]=fl.drag_sphere(Re[i])
import matplotlib.pyplot as plt
def fi(Reynolds):
    plt.loglog(Re,f)
    plt.loglog(Reynolds,fl.drag_sphere(Reynolds),'X')
    plt.xlabel('Reynolds')
    plt.ylabel('f')
    titre='Facteur de friction pour une sphère en chute libre'
    plt.title(titre)
    plt.grid(True)
    plt.show()
Cd=interactive(fi, Reynolds=(.1,100,.1));
```
